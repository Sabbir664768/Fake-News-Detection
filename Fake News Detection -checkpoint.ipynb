```python
import pandas as pd
import numpy as np

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
from collections import defaultdict
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Conv1D, GlobalMaxPooling1D, Dense, Activation
from sklearn.model_selection import train_test_split
from tensorflow.keras.layers import Flatten

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('omw-1.4')



df1 = pd.read_table('C:/Users/S_PC/Desktop/Fake Detectionall_comments.tsv',)
df1.dataframeName = 'Fake News'
nRow, nCol = df1.shape
print(f'There are {nRow} rows and {nCol} columns')
df1.head()


df2=df1.drop(['Unnamed: 0','author','parent_id','submission_id', 'ups'],axis=1)
df3 = df2[:100]


id = df3['id'].tolist()
body = df3['body'].tolist()
isTopLevel = df3['isTopLevel'].tolist()
isTopLevel



# Define the preprocessing functions
def preprocess_text(text):
    # Remove stopwords, punctuations, numbers, and multiple spaces
    stop_words = set(stopwords.words('english'))
    tokens = word_tokenize(text)
    filtered_tokens = [token.lower() for token in tokens if token.isalpha() and token.lower() not in stop_words]
#     print(filtered_tokens)
    return filtered_tokens

def lemmatize_text(tokens):
    lemmatizer = WordNetLemmatizer()
    lemmatized_tokens = [lemmatizer.lemmatize(token, get_wordnet_pos(token)) for token in tokens]
    return lemmatized_tokens

def get_wordnet_pos(token):
    # Map POS tag to first character lemmatize() accepts
    tag = nltk.pos_tag([token])[0][1][0].upper()
    tag_dict = {'J': wordnet.ADJ, 'N': wordnet.NOUN, 'V': wordnet.VERB, 'R': wordnet.ADV}
    return tag_dict.get(tag, wordnet.NOUN)

# Sample Fakeddit dataset
data = {
    'id': id,
    'body': body,
    'isTopLevel': isTopLevel
}

df = pd.DataFrame(data)

df['body'] = df['body'].fillna("")
# Preprocess the 'body' column
df['body'] = df['body'].apply(preprocess_text)
df['body'] = df['body'].apply(lemmatize_text)

# Building the vocabulary and mapping words to integer numbers
word_to_int = defaultdict(lambda: len(word_to_int) + 1)  # Assign unique integer to each word

# Transforming each text into a sequence of integers
df['body'] = df['body'].apply(lambda tokens: [word_to_int[token] for token in tokens])

# max_length = 15
# df['body'] = pad_sequences(df['body'], maxlen=max_length, padding='post', truncating='post')
# Print the preprocessed dataframe
print(df)



# Building the vocabulary and mapping words to integer numbers (same as before)
# word_to_int = defaultdict(lambda: len(word_to_int) + 1)  # Assign unique integer to each word

# Transforming each text into a sequence of integers (same as before)
# df['body'] = df['body'].apply(lambda tokens: [word_to_int[token] for token in tokens])

# Pad and truncate the sequences to a fixed length of 15 tokens
max_length = 15
df['body'] = pad_sequences(df['body'], maxlen=max_length, padding='post', truncating='post')

# Print the preprocessed and padded/truncated dataframe
# print(df)

# Generate random word embeddings
embedding_dim = 300
vocab_size = len(word_to_int) + 1
random_embeddings = np.random.rand(vocab_size, embedding_dim)



X_train, X_val, y_train, y_val = train_test_split(df['body'], df['isTopLevel'], test_size=0.2, random_state=42)


X_train = np.expand_dims(X_train, axis=-1)
X_val = np.expand_dims(X_val, axis=-1)
# for i in X_val:
#     print(i)


model = Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_length, weights=[random_embeddings], trainable=True))
model.add(Conv1D(filters=50, kernel_size=2, activation='relu'))
model.add(Conv1D(filters=50, kernel_size=3, activation='relu'))
model.add(Conv1D(filters=50, kernel_size=4, activation='relu'))
model.add(Conv1D(filters=50, kernel_size=5, activation='relu'))
model.add(GlobalMaxPooling1D())
model.add(Dense(100, activation='relu'))
model.add(Dense(2, activation='sigmoid'))


# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Print the model summary
model.summary()

# # Train the model
batch_size = 64
epochs = 10

# # model.fit(X_train, y_train, batch_size=batch_size, epochs=epochs, validation_data=(X_val, y_val))
model.fit(np.array(X_train), np.array(y_train), batch_size=batch_size, epochs=epochs, validation_data=(np.array(X_val), np.array(y_val)))
# # Evaluate the model on the validation set
loss, accuracy = model.evaluate(np.array(X_val), np.array(y_val))
print(f'Validation Loss: {loss:.4f}')
print(f'Validation Accuracy: {accuracy:.4f}')

# Predict the labels for the test set
predicted_labels = model.predict(np.array(X_val))
predicted_labels = np.round(predicted_labels).flatten()

# Create a DataFrame with the predicted labels
predicted_df = pd.DataFrame({'predicted_label': predicted_labels})

# Export the DataFrame to Excel
predicted_df.to_csv('predictions.csv', index=False)
```
